```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('../seeds/_seeds_setup/raw_streaming__subscriptions.csv')
df.head()
```

```python
staff_ids = pd.read_csv('staff_member_user_ids.csv')['user_id']
```

```python
(
    df
    .assign(fix_plan_type = lambda df: df['plan_type'].str.lower())
    .assign(staff_flag = lambda df: df['user_id'].isin(staff_ids))
.groupby(['fix_plan_type', 'monthly_fee_cents'])
.agg(
    staff = ('staff_flag', 'sum'),
    nonstaff = ('staff_flag', lambda col: col.size - col.sum()),
    all = ('staff_flag', lambda col: col.size),
)
)
```
